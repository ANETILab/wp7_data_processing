```python
import json
import polars as pl
from collections import Counter
```

```python
DIR = "/mnt/common-ssd/group-dirs/tkom-2026-project/pickwell"
CITY = "budapest"
```

```python
with open(f"{DIR}/{CITY}_old/id_lookup/id_lookup.json", "r") as fp:
    old_lookup = json.load(fp)
```

```python
with open(f"{DIR}/{CITY}/id_lookup/id_lookup.json", "r") as fp:
    new_lookup = json.load(fp)
```

```python
old = pl.DataFrame({"raw_id": old_lookup.keys(), "old_id": old_lookup.values()}).sort(
    "old_id"
)
new = pl.DataFrame({"raw_id": new_lookup.keys(), "new_id": new_lookup.values()})
new = (
    new.with_columns(
        pl.col.raw_id.str.split_exact(".", n=1).struct.rename_fields(
            ["raw_id", "aid_type"]
        )
    )
    .unnest("raw_id")
    .sort("new_id")
)
```

```python
old.join(new, on="raw_id", how="full")[["old_id", "new_id", "aid_type"]].write_csv(
    "../output/debug/old_to_new.csv"
)
```
