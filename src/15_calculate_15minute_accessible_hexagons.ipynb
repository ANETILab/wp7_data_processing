```python
from pathlib import Path

import geopandas as gpd
import yaml

with open("plotting_config.yaml", "r") as fp:
    config = yaml.safe_load(fp)
```

```python
RES = 8
AREA = "budapest"
```

```python
gzipped = ""
if RES == 10:
    gzipped = ".gz"
```

```python
walk_area = gpd.read_file(f"../data/isochrone/{AREA}/walk/15/{RES}.geojson")
hexagons = gpd.read_file(f"../data/h3/{AREA}/{RES}.geojson")
```

```python
full = walk_area.sjoin(hexagons)
full = full[["h3_left", "h3_right"]].copy()
full = full.rename(columns={"h3_left": "h3", "h3_right": "accessible"})
full = full.sort_values(["h3", "accessible"])
```

```python
Path(f"../output/{AREA}/accessible_hexagons/{RES}/").mkdir(parents=True, exist_ok=True)
```

```python
full.to_csv(
    f"../output/{AREA}/accessible_hexagons/{RES}/walk_15.csv{gzipped}", index=False
)
```
