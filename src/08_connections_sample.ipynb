```python
import polars as pl
import yaml

with open("../config.budapest.yaml", "r") as fp:
    config = yaml.safe_load(fp)
```

```python
data = pl.read_parquet(
    f"{config['dirs']['output']}/{config['city']}/hexagon_network/v2/9/connections.parquet"
)
```

```python
conn_sample = data.filter((pl.col.device_id == 6514) & (pl.col.day < "2025-06-18"))
```

```python
print(conn_sample.write_csv())
```

```python
cols = ["device_id", "day", "hour", "source", "target"]
conn_sample2 = (
    data.filter((pl.col.device_id == 6514) & (pl.col.day < "2025-06-18"))
    .group_by(cols)
    .agg(pl.col.weight.sum())
    .sort(cols)
    .with_columns(pl.col.day.str.to_date("%Y-%m-%d").dt.weekday().alias("dow"))
)
```

```python
print(conn_sample2.head(5).write_csv())
```
